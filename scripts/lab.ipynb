```python
import pandas as pd 

df = pd.read_csv("../data/annotated_data/annotated_buford.csv")


df["Category"].unique()


```

```python
# Clean up the \xa0 
df["Category"] = df["Category"].str.replace("\xa0", " ").str.strip()

# Convert all to lowercase
df["Category"] = df["Category"].str.lower()


# find sum of dialogues with category "?"
df["Category"].value_counts()
```
